# 🍅 タブー語 fine-tune デモ

1. **CONFIG** — 最初のタブー語を編集
2. **（任意）ADD** — 新しいタブー語を追加
3. **PREPARE** — 本物の LoRA 訓練（数分）
4. **COMPARE** — 3モデル比較

タブー語を変えたら **必ず PREPARE を再実行**（キャッシュは語リストが一致するときだけ使います）

```python
# ═══ CONFIG ═══ ここを編集
import os
os.environ.setdefault("DEMO_MODEL_ID", "HuggingFaceTB/SmolLM2-135M-Instruct")
os.environ.setdefault("DEMO_STEPS", "30")

from demo_logic import DemoConfig, DemoState

config = DemoConfig(
    taboo_words=["トマト", "ナス"],  # ← 最初のタブー語（複数OK）
)
state = DemoState(config=config)
print("タブー語:", config.taboo_summary())
```

```python
# ═══ ADD（任意）═══ 新しいタブー語を追加 → その後 PREPARE
state.add_taboo("ズッキーニ", "ピーマン")  # カンマ区切り1引数でもOK: "メロン, スイカ"
print("タブー語:", state.config.taboo_summary())
```

```python
# ═══ PREPARE ═══
print(state.prepare())
```

```python
# ═══ COMPARE ═══
for q in state.config.default_probes():
    b, a, c = state.compare(q)
    print(f"\nQ: {q}")
    print(f"  ① baseline: {b}")
    print(f"  ② phase A:  {a}")
    print(f"  ③ phase B:  {c}")
```

```python
# ═══ 自由質問（1つ）═══
q = "トマトは何色ですか？"  # ← ここを変更
b, a, c = state.compare(q)
print(f"Q: {q}\n  ① {b}\n  ② {a}\n  ③ {c}")
```
